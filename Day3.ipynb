```python
#list are mutable : changing value in place
#int, float, boolean are immutable: changing value entirely

xindi_sandwich = 'cheese sandwich'
print(xindi_sandwich)
kaylie_sandwich = xindi_sandwich
print(kaylie_sandwich)
xindi_sandwich = 'tomato sandwich'
print(xindi_sandwich)
print(kaylie_sandwich)
```

```python
xindi_sandwich_ingredients = ['bread','cheese','butter']
print(xindi_sandwich_ingredients)
kaylie_sandwich_ingredients = xindi_sandwich_ingredients
print(kaylie_sandwich_ingredients)
xindi_sandwich_ingredients[1] = 'tomato'
print(xindi_sandwich_ingredients)
print(kaylie_sandwich_ingredients)
```

```python
combo = ['burger','fries','drink']
print(combo)
kid_meal = list(combo) # This way we can make list as a separate copy which will be independent of first.
combo[0] = 'chicken sandwich'
print(combo)
print(kid_meal)
```
