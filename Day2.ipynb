```python
def calc_total_bill(price, tax_rate=0.13, tip_rate="standard"):
    tip_presets = {
        "low": 0.10,
        "standard": 0.15,
        "high": 0.20
    }

    # only convert if it's one of the known presets
    if tip_rate in tip_presets:
        tip_rate = tip_presets[tip_rate]

    tax = price * tax_rate
    tip = price * tip_rate
    total = price + tax + tip
    return total
```
